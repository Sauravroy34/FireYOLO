```python
data_yaml = """
path: /kaggle/input/datasets/sayedgamal99/smoke-fire-detection-yolo
train: data/train/images
val: data/val/images
test: data/test/images
nc: 2
names: ['smoke', 'fire']
"""
with open("fire_smoke.yaml", "w") as f:
    f.write(data_yaml)
print(open("fire_smoke.yaml").read())
```

```python
from ultralytics import YOLO

# Load pre-trained model (use 'n' for fastest, 's' for better accuracy)
model = YOLO("yolo26n.pt")        

# Train!
results = model.train(
    data="/kaggle/working/fire_smoke.yaml",  
    epochs=150,                               
    imgsz=640,                              
    batch=32,                                
    device=0,                                # 0 = GPU, 'cpu' if no GPU
    name="fire_smoke_yolo26n",               # folder name for runs
    patience=20,                             # early stopping
    pretrained=True,

    degrees=180,    # Random image rotation between -15 and +15 degrees
    fliplr=0.2,      # 50% chance to flip the image horizontally
    hsv_h=0.015,     # Adjust hue (color tone variations)
    hsv_s=0.7,       # Adjust image saturation
    hsv_v=0.4,
    shear=30,
    
)

```
